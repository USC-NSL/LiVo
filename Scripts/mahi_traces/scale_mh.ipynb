```python
import os
```

```python
"""
Returns a list of tuples, where each tuple is of the form (time_in_ms, freq)
"""
def generate_per_ms_freq_in_trace(path, orig_name):
    in_file = open(os.path.join(path, orig_name), "r")
    lines = in_file.readlines()
    freqs = []

    prev_time_in_ms = -1
    count = 0
    for line in lines:
        time_in_ms = int(line.split()[0])
        assert(time_in_ms >= prev_time_in_ms)

        if time_in_ms == prev_time_in_ms:
            count += 1
        else:
            if prev_time_in_ms != -1:
                freqs.append((prev_time_in_ms, count))
            count = 1
            prev_time_in_ms = time_in_ms
    
    freqs.append((prev_time_in_ms, count))
    in_file.close()
    return freqs


def scale_trace(path, orig_name, scaled_name, scale_factor):
    scale_factor = float(scale_factor)
    ms_freqs = generate_per_ms_freq_in_trace(path, orig_name)
    out_file = open(os.path.join(path, scaled_name), "w")

    for ms_freq in ms_freqs:
        freq = ms_freq[1]
        scaled_freq = freq * scale_factor
        assert(scaled_freq.is_integer())

        scaled_freq = int(scaled_freq)
        for _ in range(scaled_freq):
            out_file.write(str(ms_freq[0]) + "\n")
    out_file.close()
```

```python
path = "/home/lei/rajrup/KinectStream/Scripts/mahi_traces"
trace_names = ["mh-120-240-new"]
scale_factors = [1.5, 2.0, 2.5, 3.0, 4.0]

for trace_name in trace_names:
    for scale_factor in scale_factors:
        scale_trace(path, trace_name, trace_name + "-scaled" + str(scale_factor), scale_factor)
```
